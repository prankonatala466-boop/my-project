## 2.1 Группы по срокам финансирования

Разделим все компании на три группы:
- Единичное финансирование — был всего один раунд финансирования.
- Срок финансирования до года — между первым и последним раундом финансирования прошло не более года.
- Срок финансирования более года.

Создадим новый столбец `term`, в котором высчитаем разницу между датами `last_funding_at` и `first_funding_at`:

```python
# Рассчитываем длительность финансирования в днях
df_investments['term'] = (df_investments['last_funding_at'] - df_investments['first_funding_at']).dt.days
```

```python
# Проверим результат
df_investments['term'].head()
```

Создадим функцию для разделения всех компаний на три группы.
- 'one-off_funding' - единичное финансирование — был всего один раунд финансирования
- 'one_year_funding' - срок финансирования до года — между первым и последним раундом финансирования прошло не более года
- 'more_years_funding' - срок финансирования более года

```python
# Создаём функцию для классификации
def get_funding_group(days):
    if days == 0:
        return 'one-off_funding'
    elif 0 < days < 365:
        return 'one_year_funding'
    else:  # days >= 365
        return 'more_years_funding'

df_investments['funding_group'] = df_investments['term'].apply(get_funding_group) 
```

## 2.2 Выделение средних и нишевых сегментов рынка

Компании указывают свой сегмент рынка в столбце market. Рассчитаем, как часто в датасете встречается каждый из сегментов:

```python
# Получаем частоту каждого сегмента рынка
market_counts = df_investments['market'].value_counts()

print("Распределение сегментов рынка:")
print("=" * 60)
print(f"Всего уникальных сегментов: {len(market_counts)}")
print(f"Всего компаний: {len(df_investments):,}")
print("\nТоп-10 самых популярных сегментов:")
print(market_counts.head(10))
print("\nТоп-10 самых редких сегментов:")
print(market_counts.tail(10))
```

Сегменты, к которым относится более 120 компаний, отнесем к массовым, сегменты, в которые входит от 35 до 120 включительно, отнесем к средним, а сегменты до 35 компаний отнесем к нишевым. Рассчитаем, сколько сегментов попадает в каждую из категорий:

```python
# Определяем пороговые значения для классификации
MASSIVE_THRESHOLD = 120  # более 120 компаний - массовые
MID_THRESHOLD = 35       # от 35 до 120 включительно - средние
# менее 35 - нишевые

# Классифицируем каждый сегмент
def classify_market_segment(count):
    """
    Классифицирует сегмент рынка по количеству компаний
    
    Параметры:
    ----------
    count : int
        Количество компаний в сегменте
    
    Возвращает:
    -----------
    str
        'massive' - массовый сегмент (более 120 компаний)
        'mid' - средний сегмент (от 35 до 120 включительно)
        'niche' - нишевый сегмент (менее 35 компаний)
    """
    if count > MASSIVE_THRESHOLD:
        return 'massive'
    elif count >= MID_THRESHOLD:
        return 'mid'
    else:
        return 'niche'

# Применяем классификацию к каждому сегменту
segment_classification = market_counts.apply(classify_market_segment)

# Создаём датафрейм с информацией о сегментах
segment_df = pd.DataFrame({
    'market': market_counts.index,
    'count': market_counts.values,
    'category': segment_classification.values
})

print("\nКлассификация сегментов рынка:")
print("=" * 60)
print(segment_df.head(20))
print("=" * 60)
print(segment_df.tail(20))
```

```python
# Подсчитываем количество сегментов в каждой категории
category_counts = segment_df['category'].value_counts()

print("\nКоличество сегментов в каждой категории:")
print("=" * 60)
print(f"Массовые сегменты (> 120 компаний): {category_counts.get('massive', 0)}")
print(f"Средние сегменты (35-120 компаний): {category_counts.get('mid', 0)}")
print(f"Нишевые сегменты (< 35 компаний): {category_counts.get('niche', 0)}")
print(f"Всего сегментов: {len(segment_df)}")

# Проверяем сумму
total = category_counts.get('massive', 0) + category_counts.get('mid', 0) + category_counts.get('niche', 0)
print(f"\nСумма категорий: {total}")
print(f"Совпадает с общим количеством: {total == len(segment_df)}")
```

Массовые сегменты гарантируют финансовую стабильность и привлекательность индустрии для крупных игроков.

Средние сегменты выполняют роль эффективного «транзитного коридора», преобразуя точечные инновации в масштабные бизнесы.

***

Оставим в столбце `market` только массовые сегменты. Для остальных сегментов заменим значения на заглушки — 'niche' для нишевых и 'mid' для средних.

```python
# Создаём функцию для замены сегментов на категории
def replace_market_segment(market):
    """
    Заменяет конкретные сегменты рынка на обобщённые категории
    - массовые сегменты (> 120 компаний) оставляем без изменений
    - средние сегменты (35-120) заменяем на 'mid'
    - нишевые сегменты (< 35) заменяем на 'niche'
    """
    # Получаем количество компаний в этом сегменте
    count = market_counts.get(market, 0)
    
    if count > 120:
        return market  # массовые сегменты оставляем без изменений
    elif count >= 35:
        return 'mid'   # средние сегменты заменяем на 'mid'
    else:
        return 'niche' # нишевые сегменты заменяем на 'niche'

# Применяем функцию к столбцу market
df_investments['market_category'] = df_investments['market'].apply(replace_market_segment)

# Проверяем результат
print("Результат замены сегментов на категории:")
print("=" * 60)
print(df_investments['market_category'].value_counts())
```

```python
# Проверяем, что замена выполнена корректно
print("\nПроверка корректности замены:")
print("=" * 60)

# Проверяем, что массовые сегменты остались без изменений
massive_segments = market_counts[market_counts > 120].index.tolist()
print(f"Массовых сегментов: {len(massive_segments)}")
print(f"Примеры массовых сегментов: {massive_segments[:5]}")

# Проверяем, что средние сегменты заменены на 'mid'
mid_count = df_investments[df_investments['market_category'] == 'mid'].shape[0]
print(f"\nКомпаний в средних сегментах (mid): {mid_count:,}")

# Проверяем, что нишевые сегменты заменены на 'niche'
niche_count = df_investments[df_investments['market_category'] == 'niche'].shape[0]
print(f"Компаний в нишевых сегментах (niche): {niche_count:,}")

# Проверяем общее количество компаний в массовых сегментах
massive_count = df_investments[df_investments['market_category'].isin(massive_segments)].shape[0]
print(f"\nКомпаний в массовых сегментах: {massive_count:,}")

# Проверяем сумму
total = mid_count + niche_count + massive_count
print(f"\nВсего компаний: {total:,}")
print(f"Совпадает с общим количеством: {total == len(df_investments)}")
```

```python
# Сравниваем количество уникальных значений до и после
print("\nСравнение количества уникальных значений:")
print("=" * 60)
print(f"Уникальных сегментов до замены: {df_investments['market'].nunique()}")
print(f"Уникальных категорий после замены: {df_investments['market_category'].nunique()}")

# Показываем распределение по категориям
print("\nРаспределение по категориям:")
print(df_investments['market_category'].value_counts())

# Показываем примеры замены
print("\nПримеры замены сегментов:")
print("=" * 60)
print("До замены -> После замены")

# Примеры средних сегментов
mid_examples = market_counts[(market_counts >= 35) & (market_counts <= 120)].head(5)
for segment, count in mid_examples.items():
    print(f"  {segment} ({count} компаний) -> mid")

print("\nПримеры нишевых сегментов:")
niche_examples = market_counts[market_counts < 35].head(5)
for segment, count in niche_examples.items():
    print(f"  {segment} ({count} компаний) -> niche")
```

Определим компании с аномальным объёмом общего финансирования — используем метод IQR отдельно по каждому сегменту.

```python
# Проверяем, какие категории есть в market_category
print("Категории в market:")
print(df_investments['market_category'].value_counts().head(10))
print(f"\nВсего уникальных категорий: {df_investments['market_category'].nunique()}")
```

```python
# Функция для определения выбросов методом IQR в группе
def detect_outliers_iqr(group):
    """
    Определяет выбросы в группе методом IQR
    
    Параметры:
    ----------
    group : pandas Series
        Значения funding_total_usd для одной группы
    
    Возвращает:
    -----------
    tuple
        (нижняя_граница, верхняя_граница, маска_выбросов)
    """
    Q1 = group.quantile(0.25)
    Q3 = group.quantile(0.75)
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    # Маска выбросов (True - выброс)
    outlier_mask = (group < lower_bound) | (group > upper_bound)
    
    return lower_bound, upper_bound, outlier_mask

# Применяем функцию к каждой группе в market_category
df_investments['is_outlier_by_segment'] = False
df_investments['segment_lower_bound'] = np.nan
df_investments['segment_upper_bound'] = np.nan

# Группируем по market_category
for segment in df_investments['market_category'].unique():
    mask = df_investments['market_category'] == segment
    group_values = df_investments.loc[mask, 'funding_total_usd']
    
    if len(group_values) > 1:  # Для групп с более чем одним элементом
        lower, upper, outlier_mask = detect_outliers_iqr(group_values)
        
        df_investments.loc[mask, 'segment_lower_bound'] = lower
        df_investments.loc[mask, 'segment_upper_bound'] = upper
        df_investments.loc[mask, 'is_outlier_by_segment'] = outlier_mask

# Проверяем результат
print("Результат определения выбросов по сегментам (market_category):")
print("=" * 60)
print(f"Всего компаний: {len(df_investments):,}")
print(f"Компаний-выбросов: {df_investments['is_outlier_by_segment'].sum():,}")
print(f"Доля выбросов: {df_investments['is_outlier_by_segment'].sum() / len(df_investments) * 100:.2f}%")
    
```

```python
# Анализ выбросов по типам сегментов (massive, mid, niche)
print("\nАнализ выбросов по типам сегментов:")
print("=" * 60)

# Определяем тип сегмента
def get_segment_type(segment):
    if segment == 'mid':
        return 'mid'
    elif segment == 'niche':
        return 'niche'
    else:
        return 'massive'

df_investments['segment_type'] = df_investments['market_category'].apply(get_segment_type)

# Статистика по типам сегментов
segment_type_stats = df_investments.groupby('segment_type').agg({
    'funding_total_usd': ['count', 'mean', 'median'],
    'is_outlier_by_segment': ['sum', 'mean']
}).round(2)

segment_type_stats.columns = ['count', 'mean', 'median', 'outliers_count', 'outliers_percent']
segment_type_stats['outliers_percent'] = segment_type_stats['outliers_percent'] * 100

print(segment_type_stats)
```

## 4.1 Динамика предоставления финансирования по годам

Используя столбцы funding_total_usd и funding_rounds, для каждой компании рассчитаем средний объём одного раунда финансирования.

```python
# Рассчитываем средний объём одного раунда для каждой компании
df_final['avg_round_size'] = df_final['funding_total_usd'] / df_final['funding_rounds']

# Проверяем результат
print("Средний объём одного раунда финансирования (первые 10 компаний):")
print("=" * 60)
print(df_final[['name', 'funding_total_usd', 'funding_rounds', 'avg_round_size']].head(10))

# Статистика по среднему размеру раунда
print("\nСтатистика по среднему размеру раунда:")
print("=" * 60)
print(df_final['avg_round_size'].describe())
```

**Результаты расчёта**

Для каждой компании был рассчитан средний объём одного раунда финансирования по формуле:
`avg_round_size = funding_total_usd / funding_rounds`
**Статистика по среднему размеру раунда**

| Показатель | Значение |
|------------|----------|

1. **Медианный размер раунда составляет 1 000 000 USD**
   - Это означает, что половина компаний привлекают менее 1 млн USD за раунд
   - Это типичный размер для среднего стартапа

2. **Среднее значение (2.79 млн USD) значительно выше медианы (1 млн USD)**
   - Распределение смещено вправо

Однако есть значительное количество компаний с более крупными раундами, что создаёт широкий разброс значений. При дальнейшем анализе динамики по годам рекомендуется использовать **медианное значение**, так как оно менее чувствительно к выбросам и лучше отражает типичную ситуацию на рынке.


На основе полученных данных построим графики, которые отразят:
 - динамику типичного размера средств, которые стартапы получали в рамках одного раунда финансирования;
 - динамику общего количества раундов за каждый год, то есть насколько активно происходили инвестиции на рынке (чем больше раундов, тем выше активность).

```python
# Группируем по году финансирования (используем mid_funding_at)
yearly_stats = df_final.groupby('funding_year').agg({
    'avg_round_size': 'median',  # медианный размер раунда (типичный)
    'funding_rounds': 'sum',      # общее количество раундов
    'funding_total_usd': 'sum'    # общий объём финансирования
}).reset_index()

# Оставляем только годы с 2000 по 2014 (основной период)
yearly_stats = yearly_stats[(yearly_stats['funding_year'] >= 2000) & (yearly_stats['funding_year'] <= 2014)]

print("Статистика по годам (2000-2014):")
print("=" * 60)
print(yearly_stats)

# Создаём графики
fig, axes = plt.subplots(2, 1, figsize=(14, 10))

# 1. Динамика типичного размера раунда (медиана)
ax1 = axes[0]
ax1.plot(yearly_stats['funding_year'], yearly_stats['avg_round_size'], 
         marker='o', linewidth=2, color='#2ecc71', markersize=8)
ax1.set_xlabel('Год', fontsize=12)
ax1.set_ylabel('Медианный размер раунда (USD)', fontsize=12)
ax1.set_title('Динамика типичного размера одного раунда финансирования\n(медианное значение по годам)', fontsize=14)
ax1.grid(axis='y', alpha=0.3)
ax1.ticklabel_format(style='scientific', axis='y', scilimits=(6, 6))

# Добавляем значения на график
for x, y in zip(yearly_stats['funding_year'], yearly_stats['avg_round_size']):
    ax1.annotate(f'{y/1e6:.1f}M', (x, y), textcoords="offset points", 
                 xytext=(0, 10), ha='center', fontsize=9)

# 2. Динамика общего количества раундов
ax2 = axes[1]
ax2.bar(yearly_stats['funding_year'], yearly_stats['funding_rounds'], 
        color='#3498db', edgecolor='black', alpha=0.7)
ax2.set_xlabel('Год', fontsize=12)
ax2.set_ylabel('Количество раундов', fontsize=12)
ax2.set_title('Динамика общего количества раундов финансирования по годам', fontsize=14)
ax2.grid(axis='y', alpha=0.3)

# Добавляем значения на столбцы
for x, y in zip(yearly_stats['funding_year'], yearly_stats['funding_rounds']):
    ax2.annotate(f'{y:,.0f}', (x, y), textcoords="offset points", 
                 xytext=(0, 5), ha='center', fontsize=9)

plt.tight_layout()
plt.show()
```

## 4.2 Динамика размера общего финансирования по массовым сегментам рынка для растущих в 2014 году сегментов

## 4.3 Годовая динамика доли возвращённых средств по типам финансирования

Для каждого года и каждого вида финансирования рассчитаем нормированные значения возврата средств: то есть какую долю возвращённые средства составляют от предоставленных. При этом слишком большие аномальные значения, то есть неадекватные выбросы, заменим на пропуски.

Когда будем делить сумму возвращённых средств на суммарный объём привлечённого финансирования по конкретному году, добавим к знаменателю небольшое число, например, 1e-60. Это поможет избежать деления на ноль.

```python
# Получаем суммарный объём привлечённого финансирования по годам и типам
# Используем df_final (после всех фильтраций)
investment_by_year_type = df_final.groupby('funding_year')[funding_types].sum()

# Оставляем только годы с 2000 по 2014
investment_by_year_type = investment_by_year_type.loc[2000:2014]

print("Суммарный объём привлечённого финансирования по годам и типам:")
print("=" * 80)
print(investment_by_year_type)
```

```python
# Данные по возвратам уже в df_returns
# Проверяем структуру
print("\nДанные по возвратам (df_returns):")
print("=" * 80)
print(df_returns.head())

# Убеждаемся, что индексы - это годы
returns_by_year = df_returns.loc[2000:2014]  # берём те же годы

print("\nСуммарные возвраты по годам и типам (2000-2014):")
print("=" * 80)
print(returns_by_year)
```

```python
# Список типов финансирования для анализа
analysis_types = ['venture', 'debt_financing', 'private_equity', 'seed', 'angel']

# Создаём датафрейм для хранения нормированных значений
normalized_returns = pd.DataFrame(index=returns_by_year.index)

# Для каждого типа финансирования рассчитываем долю возвратов
for funding_type in analysis_types:
    # Получаем суммы возвратов по годам
    returns = returns_by_year[funding_type]
    
    # Получаем суммы инвестиций по годам (из df_final)
    investments = investment_by_year_type[funding_type]
    
    # Рассчитываем нормированное значение (доля возвратов)
    # Добавляем 1e-60 для избежания деления на ноль
    normalized = returns / (investments + 1e-60)
    
    # Заменяем inf и слишком большие значения (аномалии) на NaN
    # Определяем аномалии как значения > 10 (1000% возврата)
    normalized = normalized.replace([np.inf, -np.inf], np.nan)
    normalized[normalized > 10] = np.nan
    
    normalized_returns[funding_type] = normalized

print("Нормированные значения возврата средств (доля возвратов):")
print("=" * 80)
print(normalized_returns)
```

```python
# Переводим возвраты в доллары (умножаем на 1 000 000)
returns_by_year_usd = returns_by_year * 1_000_000

# Список типов финансирования для анализа
analysis_types = ['venture', 'debt_financing', 'private_equity', 'seed', 'angel']

# Создаём датафрейм для хранения нормированных значений
normalized_returns = pd.DataFrame(index=returns_by_year.index)

# Для каждого типа финансирования рассчитываем долю возвратов
for funding_type in analysis_types:
    # Получаем суммы возвратов по годам (в долларах)
    returns = returns_by_year_usd[funding_type]
    
    # Получаем суммы инвестиций по годам (из df_final, в долларах)
    investments = investment_by_year_type[funding_type]
    
    # Рассчитываем нормированное значение (доля возвратов)
    # Добавляем 1e-60 для избежания деления на ноль
    normalized = returns / (investments + 1e-60)
    
    # Заменяем inf и слишком большие значения (аномалии) на NaN
    # Определяем аномалии как значения > 10 (1000% возврата)
    normalized = normalized.replace([np.inf, -np.inf], np.nan)
    normalized[normalized > 10] = np.nan
    
    normalized_returns[funding_type] = normalized

print("Нормированные значения возврата средств (доля возвратов):")
print("=" * 80)
print(normalized_returns)

# Выводим статистику для каждого типа
print("\nСтатистика нормированных значений по типам финансирования:")
print("=" * 80)

for funding_type in analysis_types:
    data = normalized_returns[funding_type].dropna()
    if len(data) > 0:
        print(f"\n{funding_type}:")
        print(f"  - Количество лет с данными: {len(data)}")
        print(f"  - Среднее значение: {data.mean():.6f}")
        print(f"  - Медиана: {data.median():.6f}")
        print(f"  - Минимум: {data.min():.6f}")
        print(f"  - Максимум: {data.max():.6f}")
        print(f"  - Стандартное отклонение: {data.std():.6f}")
    else:
        print(f"\n{funding_type}: нет данных")
```
