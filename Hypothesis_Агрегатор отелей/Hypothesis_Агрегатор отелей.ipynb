   - Исключаем колонку `region` с помощью `.columns.difference(['region'])`.

2. **Приводим выбранные столбцы к числовому формату (`int` или `float`):**
   ```python
   df[cols_to_convert] = df[cols_to_convert].apply(pd.to_numeric, errors='coerce')

## 3. Проверка гипотез

Проверка гипотезы с одной выборкой
Туристическая индустрия является важной частью экономики, и анализ её показателей позволяет выявить тренды и оценить эффективность работы рынка. Один из ключевых показателей — средняя стоимость реализованных турпакетов.

Существует мнение, что средняя стоимость реализованных турпакетов в 2023 году составила 800 млн рублей. Однако аналитики подозревают, что этот показатель может быть больше. Задача — провести статистический тест, который поможет определить, можно ли отвергнуть нулевую гипотезу в пользу альтернативной.

Нужно проверить одностороннюю гипотезу:

- **H0**: Средняя стоимость реализованных турпакетов в 2023 году (млн рублей) = 800
- **H1**: Средняя стоимость реализованных турпакетов в 2023 году (млн рублей) > 800

````python
# Выборка — общая стоимость турпакетов в 2023 году по регионам
sample = df_tour_cost_2023['total_tour_cost_million_2023'].dropna()

# Базовая статистика
print(f'Размер выборки: {len(sample)}')
print(f'Среднее: {sample.mean():.2f}')
print(f'Медиана: {sample.median():.2f}')
print(f'Стандартное отклонение: {sample.std(ddof=1):.2f}')
print(f'Минимум: {sample.min():.2f}')
print(f'Максимум: {sample.max():.2f}')
````

Проверка условий применимости t-теста

t-тест корректно применять, если:

наблюдения независимы;
распределение примерно нормальное или размер выборки достаточно большой (n > 30).

````python
# Визуальная проверка
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
sns.histplot(sample, kde=True, bins=20)
plt.axvline(800, color='red', linestyle='--', label='H0: 800')
plt.title('Распределение стоимости турпакетов')
plt.xlabel('Стоимость, млн руб.')
plt.legend()

plt.subplot(1, 2, 2)
stats.probplot(sample, dist='norm', plot=plt)
plt.title('Q-Q plot')

plt.tight_layout()
plt.show()

# Тест Шапиро — Уилка (H0: распределение нормальное)
shapiro_stat, shapiro_p = stats.shapiro(sample)
print(f'Shapiro-Wilk: statistic = {shapiro_stat:.4f}, p-value = {shapiro_p:.4f}')
````

````python
# Одновыборочный t-тест
# alternative='greater' — проверяем H1: среднее > 800
t_stat, p_value = stats.ttest_1samp(sample, popmean=800, alternative='greater')

print(f't-статистика: {t_stat:.4f}')
print(f'p-value (односторонний): {p_value:.4f}')

alpha = 0.05
if p_value < alpha:
    print(f'p-value < {alpha}: отвергаем H0 в пользу H1')
    print('Средняя стоимость турпакетов статистически значимо больше 800 млн руб.')
else:
    print(f'p-value >= {alpha}: нет оснований отвергнуть H0')
    print('Средняя стоимость турпакетов статистически значимо не отличается от 800 млн руб.')
````

````python
# Односторонний 95% ДИ (нижняя граница)
confidence = 0.95
n = len(sample)
mean = sample.mean()
se = sample.std(ddof=1) / np.sqrt(n)
t_crit = stats.t.ppf(confidence, df=n-1)

lower_bound = mean - t_crit * se
print(f'Среднее: {mean:.2f}')
print(f'Нижняя граница 95% ДИ: {lower_bound:.2f} млн руб.')
print(f'Верхняя граница: +бесконечность')
````

## Проблемы и ограничения

1. **Сильная асимметрия распределения.** Медиана (2 599) в 3,3 раза меньше среднего (8 579). Это значит, что среднее сильно «перекошено» вправо из-за Москвы.

2. **Наличие выброса.** Москва с показателем 388 216 млн руб. — это более чем в 45 раз больше среднего по остальным регионам. Такой выброс оказывает огромное влияние на t-статистику и p-value.

3. **Условие нормальности не выполнено.** Формально t-тест предполагает нормальность или достаточно большой объем выборки, чтобы опираться на ЦПТ. При n = 85 и таком выбросе ЦПТ работает плохо.

4. **Хрупкость результата.** p-value = 0.0456 очень близко к порогу 0.05. Небольшое изменение выборки (например, исключение одного выброса) может изменить вывод.


````python
# Проведем тест без Москвы
sample_no_msk = sample[sample < 100000]  # убираем Москву
t_stat2, p_value2 = stats.ttest_1samp(sample_no_msk, popmean=800, alternative='greater')
print(f'Без Москвы: t = {t_stat2:.4f}, p = {p_value2:.4f}')
````

## Вывод по проверке гипотезы о средней стоимости турпакетов

- **С полной выборкой (85 регионов):** t = 1.7092, p = 0.0456 — H0 отвергается, но результат на грани порога 0.05.
- **Без выброса (Москвы, 84 региона):** t = 5.9853, p = 0.0000 — H0 уверенно отвергается.

Исключение Москвы не ослабило, а усилило результат. Это значит, что средняя стоимость турпакетов в 2023 году статистически значимо превышает 800 млн руб. не только за счет столицы, но и за счет остальных регионов.

**Итог:** гипотеза H0 отвергается в пользу H1 — средняя стоимость реализованных турпакетов в 2023 году значимо больше 800 млн руб. Результат устойчив и подтверждается на выборке без выбросов.

---

**Проверка гипотезы с двумя выборками**

Статистика говорит, что в 2023 году, по сравнению с 2022, россияне стали чаще ездить по стране и оформлять турпакеты. Проверим статистическую гипотезу о том, стало ли в 2023 году больше проданных турпакетов по России, чем в 2022 году.

Проверим одностороннюю гипотезу:

- **H0**: Среднее количество турпакетов, реализованных в 2022 году = Среднему количеству турпакетов, реализованных в 2023 году
- **H1**: Среднее количество турпакетов, реализованных в 2022 году < Cреднего количества турпакетов, реализованных в 2023 году

````python
# Формируем парные выборки
sample_2022 = df_tour_firm['domestic_russia_2022']
sample_2023 = df_tour_firm['domestic_russia_2023']

# Убираем возможные пропуски (если есть)
mask = sample_2022.notna() & sample_2023.notna()
sample_2022 = sample_2022[mask]
sample_2023 = sample_2023[mask]

print(f'Размер парной выборки: {len(sample_2022)}')
print(f'Среднее 2022: {sample_2022.mean():.2f}')
print(f'Среднее 2023: {sample_2023.mean():.2f}')
print(f'Разница средних (2023 − 2022): {sample_2023.mean() - sample_2022.mean():.2f}')
````

````python
# Разности пар
diff = sample_2023 - sample_2022

# Визуальная проверка
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
sns.histplot(diff, kde=True, bins=20)
plt.axvline(0, color='red', linestyle='--', label='H0: разница = 0')
plt.title('Распределение разностей (2023 − 2022)')
plt.xlabel('Разница числа турпакетов')
plt.legend()

plt.subplot(1, 2, 2)
stats.probplot(diff, dist='norm', plot=plt)
plt.title('Q-Q plot разностей')

plt.tight_layout()
plt.show()

# Тест Шапиро — Уилка для разностей
shapiro_stat, shapiro_p = stats.shapiro(diff)
print(f'Shapiro-Wilk (разности): statistic = {shapiro_stat:.4f}, p = {shapiro_p:.4f}')
````

````python
# Парный t-тест
# (2023, 2022),  alternative='greater'
t_stat, p_value = stats.ttest_rel(sample_2023, sample_2022, alternative='greater')

print(f't-статистика: {t_stat:.4f}')
print(f'p-value (односторонний): {p_value:.4f}')

alpha = 0.05
if p_value < alpha:
    print(f'p-value < {alpha}: отвергаем H0 в пользу H1')
    print('В 2023 году турпакетов по России стало статистически значимо больше, чем в 2022.')
else:
    print(f'p-value >= {alpha}: нет оснований отвергнуть H0')
    print('Статистически значимого роста числа турпакетов по России не обнаружено.')
````

### Что было сделано

Проведен парный t-тест Стьюдента (`ttest_rel`) для проверки гипотезы о росте числа турпакетов, реализованных гражданам России по территории России, в 2023 году по сравнению с 2022 годом. Использованы колонки `domestic_russia_2022` и `domestic_russia_2023`.

| p-value (односторонний) | 0.0962 |

### Проверка условий

Тест Шапиро — Уилка (p = 0.0000) и Q-Q plot показали, что разности **не распределены нормально**. Причина — крупный выброс (~220 000 турпакетов). Формально условие применимости парного t-теста не выполнено.

### Результат

p-value = 0.0962 > 0.05 — **нет оснований отвергнуть H0** на уровне значимости 0.05. Статистически значимого роста числа турпакетов по России в 2023 году по сравнению с 2022 годом не обнаружено.

### Ограничения

- Разности не нормальны — формально парный t-тест неприменим.
- Наличие крупного выброса сильно влияет на результат.
- p-value близко к порогу 0.05 — результат неустойчив.

### Итог

Формально гипотеза H0 **не отвергается**: статистически значимого роста числа турпакетов по России в 2023 году не обнаружено. Однако наблюдается положительная тенденция (+10% к среднему), которая не достигла уровня значимости из-за высокой вариативности данных и наличия выброса. Для окончательного вывода требуется дополнительный анализ с использованием непараметрических методов.

Применим еще  непараметрический тест Уилкоксона

````python
wilcoxon_stat, wilcoxon_p = stats.wilcoxon(sample_2023, sample_2022, alternative='greater')
print(f'Wilcoxon: statistic = {wilcoxon_stat:.4f}, p = {wilcoxon_p:.4f}')
````

Формально статистически значимого роста числа турпакетов по России в 2023 году по сравнению с 2022 годом не обнаружено (парный t-тест: p = 0.0962; тест Уилкоксона: p = 0.0618).
